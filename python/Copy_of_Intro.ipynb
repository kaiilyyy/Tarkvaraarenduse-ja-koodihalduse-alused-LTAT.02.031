`print()`, `type()` ja `len()` on **sisseehitatud funktsioonid**: need on kohe kasutatavad, midagi pole vaja importida ega paigaldada. [Sisseehitatud funktsioonide teatmikust](https://docs.python.org/3.12/library/functions.html) leiad kogu loendi koos parameetrite ja näidetega. Omistamine võib kasutada varasemat väärtust: kõigepealt arvutatakse parem pool ja seejärel seotakse nimi uue väärtusega.

Arvulise muutuja väärtuse muutmiseks on ka lühivormid ehk **liitomistamised**. Näiteks `+=` liidab olemasolevale väärtusele paremal antud arvu ja salvestab tulemuse samasse muutujasse. Eelmise näite rea `vanus = vanus + 1` võid seega kirjutada lühemalt: `vanus += 1`.

| Lühivorm | Pikem kuju arvulise muutuja korral | Tegevus |
| --- | --- | --- |
| `arv += 2` | `arv = arv + 2` | Suurendab väärtust kahe võrra |
| `arv -= 2` | `arv = arv - 2` | Vähendab väärtust kahe võrra |
| `arv *= 2` | `arv = arv * 2` | Korrutab väärtuse kahega |
| `arv /= 2` | `arv = arv / 2` | Jagab väärtuse kahega |

Need käsud muudavad muutuja väärtust. Võrdlusmärke `==` ja `!=` vaatame peatükis [I4](#i4).

Arvutustes saad kasutada märke `+`, `-`, `*`, `/` ja `**` (astendamine). `%` annab jagamisjäägi. `len()` annab kogumi või sõne pikkuse. Väärtuse teisendamiseks teise tüüpi saad kasutada näiteks `int()`, `float()` ja `str()`.

**F-sõne** korral on sõne alustava jutumärgi ees täht `f`. Loogelistesse sulgudesse saad kirjutada muutuja või arvutuse, mille tulemus lisatakse teksti. Vorming `:.2f` kuvab arvu kahe komakohaga.

Dokumentatsioon: [arvud ja sõned](https://docs.python.org/3.12/tutorial/introduction.html#using-python-as-a-calculator), [f-sõned](https://docs.python.org/3.12/tutorial/inputoutput.html#formatted-string-literals).

````python
# F-sõne: arvutuse tulemuse lisamine teksti kahe komakohaga.
hind = 2.50
kogus = 3
print(f"{kogus} eset maksavad {hind * kogus:.2f} eurot.")

# Tüübi teisendamine: sõnest täisarvuks.
print(int("12") + 1)

# Sõne pikkuse leidmine.
print(len("Python"))

# Jagamisjäägi leidmine.
print(7 % 2)
````

## I3. Andmestruktuurid, funktsioonid ja meetodid

**List** (`list`) hoiab väärtusi kindlas järjekorras. Nurksulgudes olev indeks näitab elemendi asukohta listis: esimese elemendi indeks on `0`.

Mitme järjestikuse elemendi valimiseks kasuta **lõiku** (*slice*). Näiteks `arvud[1:3]` tagastab uue listi elementidega indeksitel 1 ja 2 ehk esialgse listi teise ja kolmanda elemendiga. Valik algab esimesest antud indeksist ja lõpeb enne teist: indeksiga 3 elementi kaasa ei võeta.

Listi saad anda argumendiks ka sisseehitatud funktsioonile: `min(arvud)`, `max(arvud)` ja `sum(arvud)` tagastavad vastavalt väikseima väärtuse, suurima väärtuse ja summa. Need ei muuda listi.

**Meetod** on objektiga seotud funktsioon. Meetodi väljakutses kirjuta objekti tähistava nime järele punkt, meetodi nimi ja sulud: `arvud.append(12)`. Listi meetod `append()` muudab olemasolevat listi, lisades lõppu väärtuse. See tagastab `None`, seega ära kirjuta `arvud = arvud.append(12)`.

Võrdle kirjapilti: `sum(arvud)` kutsub funktsiooni, `arvud.append(12)` listi meetodit. Kõik meetodid ei muuda objekti: näiteks sõne meetod `"Mari".upper()` tagastab uue sõne `"MARI"`.

Dokumentatsioon: [listi meetodid](https://docs.python.org/3.12/tutorial/datastructures.html#more-on-lists), [min, max ja sum](https://docs.python.org/3.12/library/functions.html).

````python
# Listi loomine ja esimese elemendi valimine indeksiga.
arvud = [4, 7, 9, 10]
print(arvud[0])

# Lõik: teise ja kolmanda elemendi valimine.
print(arvud[1:3])

# Meetod append(): olemasoleva listi muutmine.
arvud.append(12)
print(arvud)

# Sisseehitatud funktsioonid: väikseim väärtus, suurim väärtus ja summa.
print(min(arvud), max(arvud), sum(arvud))
````

Omistus `teine = algne` annab samale listile teise nime. Kui muudad listi ühe nime kaudu, on muutus näha ka teise kaudu. `.copy()` loob eraldi listi. Järgmise lihtsa arvulisti koopiat saad muuta algset listi mõjutamata; kui list sisaldab teisi liste, jagab selline koopia algse listiga neid sisemisi liste.

````python
# Omistus: mõlemad nimed viitavad samale listile.
algne = [4, 7]
teine = algne
teine.append(9)
print(algne)

# Koopia: selle listi muutmine ei muuda algset listi.
koopia = algne.copy()
koopia.append(12)
print(algne, koopia)
````

**Sõnastik** (`dict`) hoiab võtme ja väärtuse paare. Väärtuse saamiseks kasuta võtit, mitte järjekorranumbrit. Sõnastik säilitab võtmete lisamise järjekorra; samale võtmele uue väärtuse määramine asendab varasema väärtuse.

Dokumentatsioon: [sõnastikud](https://docs.python.org/3.12/tutorial/datastructures.html#dictionaries).

````python
# Sõnastiku loomine ja väärtuse valimine võtme järgi.
osaleja = {"nimi": "Mari", "vanus": 24}
print(osaleja["nimi"])

# Uue võtme ja väärtuse paari lisamine.
osaleja["linn"] = "Tartu"
print(osaleja)
````

### Põhiliste andmestruktuuride ülevaade

Järgmisi tüüpe tasub koodi lugedes ära tunda. Tabeli viimane veerg näitab, kas olemasoleva kogumi elemente saab asendada, lisada või eemaldada.

| Tüüp | Näide | Järjekord ja valimine | Kas kogumit saab muuta? |
| --- | --- | --- | --- |
| List (`list`) | `[4, 7, 4]` | Järjestatud; valik indeksiga | Jah |
| Ennik (`tuple`) | `(4, 7)` | Järjestatud; valik indeksiga | Ei |
| Sõnastik (`dict`) | `{"nimi": "Mari", "vanus": 24}` | Säilitab lisamise järjekorra; valik võtmega | Jah |
| Hulk (`set`) | `{"A", "T", "S"}` | Kordumatud elemendid; kindlat järjekorda ega indeksit pole | Jah |
| Muutumatu hulk (`frozenset`) | `frozenset({"A", "T"})` | Kordumatud elemendid; kindlat järjekorda ega indeksit pole | Ei |
| Täisarvujada (`range`) | `range(1, 4)` | Järjestatud; siin väärtused 1, 2 ja 3 | Ei |

Tühi `{}` on sõnastik; tühja hulga loob `set()`. Ühe elemendiga ennikus on koma: `(4,)`. Ennik sobib näiteks koordinaatide hoidmiseks või funktsioonist mitme väärtuse korraga tagastamiseks. Enniku elemente ei saa asendada, kuid selle sees olevat listi saab siiski muuta.

Hulk sobib kordumatute väärtuste leidmiseks ja kontrollimiseks, kas väärtus kuulub hulka. `range` kirjeldab täisarvude jada ilma kõiki arve listi salvestamata; kasutad seda tsüklite juures. Ka sõne on järjestatud ja muutumatu: `"Mari"[0]` annab `"M"`.

Dokumentatsioon: [ennikud](https://docs.python.org/3.12/tutorial/datastructures.html#tuples-and-sequences), [hulgad](https://docs.python.org/3.12/tutorial/datastructures.html#sets), [set ja frozenset](https://docs.python.org/3.12/library/stdtypes.html#set-types-set-frozenset), [range](https://docs.python.org/3.12/library/stdtypes.html#ranges).

````python
# Ennik: väärtuste hoidmine koos ja omistamine eri muutujatele.
koordinaadid = (58.38, 26.72)
laius, pikkus = koordinaadid  # Esimene väärtus omistatakse muutujale laius, teine muutujale pikkus.
print(laius, pikkus)

# Hulk: kordumatute väärtuste leidmine.
ravimikoodid = ["A", "T", "A", "S"]
unikaalsed = set(ravimikoodid)
print(len(unikaalsed))

# Hulka kuulumise kontrollimine.
print("A" in unikaalsed)

# Hulga väärtustest järjestatud listi loomine.
print(sorted(unikaalsed))  # sorted() tagastab järjestatud listi.
````

## I4. Võrdlused ja tingimuslause

`==` kontrollib, kas väärtused on võrdsed; `!=` kontrollib, kas need on erinevad. Üksik `=` omistab väärtuse. Teised võrdlusmärgid on `<`, `<=`, `>` ja `>=`. Võrdluse tulemus on `True` või `False`. `in` kontrollib, kas väärtus kuulub kogumisse.

Tingimusi saab ühendada operaatoritega `and` (ja), `or` (või) ja `not` (eitus). `not in` kontrollib, kas väärtus puudub kogumist.

`is` kontrollib, kas kaks nime viitavad samale objektile; `==` võrdleb väärtusi. Väärtuse puudumist kontrolli kujul `tulemus is None` või `tulemus is not None`. Arvude ja sõnede võrdlemiseks kasuta `==`.

````python
# Kahe võrdluse ühendamine operaatoriga and.
vanus = 24
print(vanus >= 18 and vanus < 65)

# Listi kuulumise kontroll operaatoriga in.
print("Tartu" in ["Tallinn", "Tartu"])

# or ja not: vähemalt üks tingimus kehtib; tingimuse eitamine.
print(vanus < 18 or vanus >= 65)
print(not (vanus < 18))

# not in: väärtus puudub listist.
print("Pärnu" not in ["Tallinn", "Tartu"])

# is None: väärtuse puudumise kontroll.
tulemus = None
print(tulemus is None)
````

Programmi **täitmise juhtimine** (*control flow*) määrab, milliseid lauseid ja mis järjekorras täidetakse. Tavaliselt liigub täitmine ülalt alla; tingimus valib haru ja tsükkel kordab plokki.

`if` käivitab ploki, kui tingimus kehtib. `elif` kontrollib järgmist tingimust ainult siis, kui eelmine ei kehtinud. `else` käivitub, kui ükski eelnev tingimus ei kehtinud. Ühes `if`/`elif`/`else` ahelas täidetakse kõige rohkem üks haru; mitu eraldi `if`-lauset kontrollivad igaüks oma tingimust.

Tingimuse rea lõpus on koolon. Sellele järgnevad ploki read kirjuta **taandega**, lisades rea algusesse neli tühikut. Plokk lõpeb seal, kus järgmise koodirea taane on väiksem. Ära sega tabulaatoreid ja tühikuid.

Dokumentatsioon: [võrdlused ja loogika](https://docs.python.org/3.12/library/stdtypes.html#boolean-operations-and-or-not), [if-laused](https://docs.python.org/3.12/tutorial/controlflow.html#if-statements).

````python
# Tingimuslause: käivitub ainult sobiva haru taandega kood.
temperatuur = 3
if temperatuur < 0:
    print("Alla nulli")
elif temperatuur == 0:
    print("Null kraadi")
else:
    print("Üle nulli")

# Taandeta rida jääb tingimuslausest välja ja käivitub iga haru järel.
print("Kontroll lõppes.")
````

## I5. Tsüklid ja nende juhtimine

`for` läbib kogumi elemente ükshaaval. Tsükli muutuja saab igal sammul järgmise elemendi väärtuse. Tsükli sees olevat taandega koodi täidetakse iga elemendi jaoks.

Järgmine näide kogub negatiivsed mõõtetulemused uude listi. Tühi tulemuste list luuakse enne tsüklit, et varasemad tulemused iga sammu alguses ei kaoks. `print()` jääb tsüklist välja ja käivitub üks kord. Ennusta tulemust enne käivitamist.

````python
mootmised = [-2, 4, -1, 0, 5]
negatiivsed = []

# for ja if: tingimusele vastavate väärtuste kogumine uude listi.
for mootmine in mootmised:
    if mootmine < 0:
        negatiivsed.append(mootmine)

# Pärast tsüklit: kogu tulemuse ja selle pikkuse kuvamine.
print(negatiivsed)
print(len(negatiivsed))
````

Esimesel sammul lisatakse `-2`, teisel jääb `4` lisamata ja kolmandal lisatakse `-1`. Ka `0` ja `5` jäävad lisamata. Kui listi poleks ühtegi väärtust lisatud, oleks selle pikkus `0`.

`range(1, 4)` annab arvud 1, 2 ja 3; lõppväärtus jääb välja. `enumerate(kogum, start=1)` annab iga väärtusega kaasa järjekorranumbri. Sõnastiku `.items()` annab võtme ja väärtuse paarid. Kahte tsüklimuutujasse jaotatakse paari väärtused nagu enniku lahtipakkimisel peatükis I3.

Dokumentatsioon: [for ja range](https://docs.python.org/3.12/tutorial/controlflow.html#for-statements), [kogumite läbimine](https://docs.python.org/3.12/tutorial/datastructures.html#looping-techniques).

````python
# range(): arvude jada läbimine.
for number in range(1, 4):
    print(number)

# enumerate(): järjekorranumber koos väärtusega.
for number, mootmine in enumerate([-2, 4, 5], start=1):
    print(number, mootmine)

# items(): sõnastiku võtmed koos väärtustega.
for nimi, vaartus in {"väikseim": -2, "suurim": 5}.items():
    print(nimi, vaartus)
````

### Millal korrata, vahele jätta või lõpetada?

`for` sobib kogumi läbimiseks. **`while` kordab plokki seni, kuni tingimus on tõene**; tingimust kontrollitakse enne iga sammu. Näites vähendame loendurit, kuni see jõuab nulli. Kui tingimus jääb alati tõeseks, tekib lõputu tsükkel. Selle saad katkestada töövihiku nupuga **Stop/Interrupt** või terminalis klahvidega **Ctrl + C**.

**`continue`** jätab käimasoleva sammu ülejäänud käsud vahele ja liigub järgmisele sammule. **`break`** lõpetab lähima ümbritseva tsükli; täitmine jätkub pärast seda tsüklit. Allpool jätame tühja sõne vahele ja lõpetame märgusõna `"STOP"` juures.

Koodis võib ette tulla ka `pass`: see ei tee midagi ja sobib veel täitmata ploki kohatäiteks. See ei asenda `continue`- ega `break`-lauset.

Dokumentatsioon: [while](https://docs.python.org/3.12/tutorial/introduction.html#first-steps-towards-programming), [break ja continue](https://docs.python.org/3.12/tutorial/controlflow.html#break-and-continue-statements).

````python
# while: tingimus vaadatakse enne iga sammu uuesti üle.
kordi = 3
while kordi > 0:
    print(kordi)
    kordi -= 1
print("Tsükkel lõppes.")
````

````python
for tekst in ["4", "", "5", "STOP", "9"]:
    # continue: tühi sõne jääb kuvamata.
    if tekst == "":
        continue
    # break: STOP-i ja sellele järgnevaid väärtusi ei kuvata.
    if tekst == "STOP":
        break
    print(tekst)
print("Lugemine lõppes.")
````

## I6. Funktsioon ja tagastusväärtus

Funktsioon võimaldab tegevuse ühe korra kirja panna ja seda eri väärtustega korrata. Funktsiooni defineerimisel kirjuta `def` järele funktsiooni nimi ning sulgudesse **parameetrid**. Parameetrid on nimed, mille kaudu funktsioon kasutab talle antud väärtusi. Väljakutsel antud väärtusi nimetatakse **argumentideks**. Funktsiooni sees olev kood käivitub alles siis, kui funktsiooni välja kutsud.

`return` lõpetab funktsiooni täitmise kohe ja tagastab tulemuse väljakutse kohta. Tsükli sees olev `return` lõpetab ka funktsiooni, mitte ainult tsükli. `print()` ainult kuvab väärtuse; see ei asenda tagastamist. Kui funktsiooni täitmine jõuab lõpuni ilma `return`-lauseta, tagastab funktsioon `None`.

Dokumentatsioon: [funktsiooni defineerimine](https://docs.python.org/3.12/tutorial/controlflow.html#defining-functions).

Järgmine funktsioon saab mõõtmised ja alumise piiri. `alampiir=0` määrab vaikeväärtuse: kui jätad teise argumendi andmata, kasutatakse nulli. Tagastame kaks väärtust ennikuna: sobivate mõõtmiste arvu ja summa. Väljakutsel saad argumendi anda asukoha järgi või parameetri nimega, näiteks `alampiir=4`.

````python
# Funktsiooni kirjeldus: sisend, vaikeväärtus ja enniku tagastamine.
def mootmiste_kokkuvote(arvud, alampiir=0):
    """Tagasta vähemalt alampiiriga võrdsete mõõtmiste arv ja summa."""
    valitud = []
    for arv in arvud:
        if arv >= alampiir:
            valitud.append(arv)
    return len(valitud), sum(valitud)

# Asukoha järgi antud argument; alampiiriks jääb vaikeväärtus 0.
arv, summa = mootmiste_kokkuvote([-2, 4, -1, 0, 5])
print(arv, summa)

# Nimega argument: sama funktsiooni kasutamine teise piiriga.
arv, summa = mootmiste_kokkuvote([-2, 4, -1, 0, 5], alampiir=4)
print(arv, summa)
````

`valitud` on funktsiooni kohalik muutuja: väljaspool funktsiooni ei saa seda nime otse kasutada. Funktsiooni arvutatud tulemuse saad kätte `return` abil. Funktsiooni sisend anna argumentidena, et see ei sõltuks mõnes varasemas lahtris loodud mõõtmiste listist. Seevastu `if`- või `for`-plokis loodud muutuja on kasutatav ka pärast ploki lõppu, kui muutuja loonud rida käivitati.

Tulemuse kontrollimiseks saad kasutada `assert`-lauset. Kui tingimus on tõene, ei kuvata midagi. Väära tingimuse korral tekib `AssertionError`. Kasuta näiteid, mille vastust tead ette.

Dokumentatsioon: [assert](https://docs.python.org/3.12/reference/simple_stmts.html#the-assert-statement).

````python
# Tavaline sisend, piirväärtus ja tühi sisend.
assert mootmiste_kokkuvote([-2, 4, -1, 0, 5]) == (3, 9)
assert mootmiste_kokkuvote([3, 4, 5], alampiir=4) == (2, 9)
assert mootmiste_kokkuvote([]) == (0, 0)
print("Kontrollid õnnestusid.")
````

### Failitee ja faili sisu

`Path` tähistab failiteed. `Path.cwd()` näitab töökataloogi: suhteline tee `valjund/mootmised.txt` algab sellest kaustast, mitte tingimata skripti asukohast. `kaust / "mootmised.txt"` ühendab tee osad; `mkdir()` loob kausta ja `resolve()` näitab täielikku teed. Atribuut `fail.name` annab failinime ilma meetodi väljakutse sulgudeta.

`fail` on siin failiteed kirjeldav `Path`-objekt. `fail.open(...)` tagastab **avatud faili objekti**, millele annab `as fh` nime `fh` (lühend sõnadest *file handle*). Selle objekti kaudu loeme või kirjutame sisu; objekt peab järge ka lugemis- või kirjutamiskoha üle. `fh` on meie valitud muutujanimi, mitte Pythoni märksõna.

Faili avamisel kasuta `with`-plokki: plokist väljumisel suletakse fail ka erindi korral. Režiim `"w"` loob faili või kirjutab varasema sisu üle, `"a"` lisab teksti lõppu ja `"r"` loeb faili. `encoding="utf-8"` sobib eestikeelse teksti jaoks.

Faili kirjutamiseks kasuta `print(..., file=fh)`: `file` suunab väljundi avatud faili. Vaikimisi eraldab `print()` väärtused tühikuga ja lisab reavahetuse. `sep` muudab eraldajat, `end` rea lõppu. Faili meetod `write()` kirjutab valmis sõne ega lisa ise reavahetust.

Loome väikese mõõtmiste faili. Igal real on üks arv; tühi rida ja tekst `puudub` võimaldavad proovida puuduliku sisendi käsitlemist. Kogu näite korduskäivitus taastab sama faili, sest esimene kirjutamine kasutab režiimi `"w"`.

Dokumentatsioon: [Path](https://docs.python.org/3.12/library/pathlib.html#basic-use), [failide lugemine ja kirjutamine](https://docs.python.org/3.12/tutorial/inputoutput.html#reading-and-writing-files).

````python
from pathlib import Path

# Failitee: valjund luuakse praegusesse töökataloogi.
print("Töökataloog:", Path.cwd())
kaust = Path("valjund")
kaust.mkdir(exist_ok=True)
fail = kaust / "mootmised.txt"

# Režiim w: faili loomine print() abil, üks väärtus real.
with fail.open("w", encoding="utf-8") as fh:
    for tekst in ["-2", "4", "", "puudub", "5"]:
        print(tekst, file=fh)

# Režiim a: kahe mõõtmise lisamine faili lõppu.
with fail.open("a", encoding="utf-8") as fh:
    print("-1", "0", sep="\n", file=fh)

# Režiim r ja read(): kogu teksti lugemine ning kuvamine.
with fail.open("r", encoding="utf-8") as fh:
    print(fh.read(), end="")
print("Fail:", fail.name, "Asukoht:", fail.resolve())
````

### Mõõtmiste lugemine ja kokkuvõte

Avatud tekstifaili objekt on **itereeritav**: `for rida in fh` annab igal sammul järgmise rea sõnena, kogu faili enne listi kogumata. `enumerate(fh, start=1)` lisab loenduri alates ühest. Kohe pärast faili avamist vastab see reanumbrile, ka tühjad read loetakse kaasa.

Näites eemaldame `.strip()` abil tühimärgid, jätame tühja rea vahele ja proovime teisendada teksti arvuks. Vigase väärtuse korral kuvame reanumbriga teate ja jätkame järgmise reaga.

Funktsioon `loe_mootmised` tagastab arvude listi. Anname selle peatükis I6 kirjutatud funktsioonile `mootmiste_kokkuvote`. Nii on faili lugemine ja arvutamine eraldi: sama kokkuvõttefunktsioon töötab ka käsitsi koostatud listiga.

Dokumentatsioon: [failiobjekti meetodid ja ridade läbimine](https://docs.python.org/3.12/tutorial/inputoutput.html#methods-of-file-objects).

````python
# Faili lugemine: tühjade ridade ja teisendusvea käsitlemine.
def loe_mootmised(failitee):
    """Loe arvud tekstifailist; jäta tühjad ja vigased read vahele."""
    arvud = []
    with failitee.open("r", encoding="utf-8") as fh:
        for reanumber, rida in enumerate(fh, start=1):
            tekst = rida.strip()
            # Tühja rea vahelejätmine.
            if tekst == "":
                continue
            # Teksti teisendamine ja vigase rea käsitlemine.
            try:
                arv = float(tekst)
            except ValueError:
                print(f"Rida {reanumber} jäi vahele: {tekst}")
                continue
            arvud.append(arv)
    return arvud

# Kahe funktsiooni ühendamine: esimese tulemus on teise sisend.
loetud = loe_mootmised(fail)
arv, summa = mootmiste_kokkuvote(loetud, alampiir=0)
print("Arv:", arv, "Summa:", summa)

# Kokkuvõtte salvestamine.
with (kaust / "kokkuvote.txt").open("w", encoding="utf-8") as fh:
    print("Arv:", arv, file=fh)
    print("Summa:", summa, file=fh)
````

## I9. Lisamaterjal: objekt ja klass

Listid, sõnastikud ja andmetabelid on objektid. **Klass** kirjeldab, millised andmed ja meetodid teatud tüüpi objektil on. Meetodite kasutamiseks ei pea oskama uusi klasse kirjutada.

Järgmises näites määrab `__init__` uue ristküliku laiuse ja kõrguse. `self` viitab objektile endale. Laius ja kõrgus on atribuudid ning `pindala()` on meetod.

Dokumentatsioon: [klassid ja objektid](https://docs.python.org/3.12/tutorial/classes.html#a-first-look-at-classes).

````python
# Klassi kirjeldus: objekti atribuudid ja meetod.
class Ristkulik:
    def __init__(self, laius, korgus):
        self.laius = laius
        self.korgus = korgus

    def pindala(self):
        return self.laius * self.korgus

# Objekti loomine selle klassi põhjal.
kujund = Ristkulik(3, 4)

# Atribuudi väärtuse lugemine.
print(kujund.laius)

# Meetodi kutsumine pindala arvutamiseks.
print(kujund.pindala())
````
